```python
def two_sum(nums, target):
    seen = {}

    for i, num in enumerate(nums):
        complement = target - num

        if complement in seen:
            return [seen[complement], i]

        seen[num] = i

    return [-1, -1]


def main():
    test_cases = [
        ([2, 7, 11, 15], 9),
        ([3, 2, 4], 6),
        ([3, 3], 6),
        ([1, 2, 3], 10),
        ([], 5),
        ([5], 10),
        ([-3, 4, 3, 90], 0),
        ([0, 0], 0),
    ]

    for nums, target in test_cases:
        result = two_sum(nums, target)

        print(
            f"nums={nums}, "
            f"target={target}, "
            f"result={result}"
        )


if __name__ == "__main__":
    main()
```

```python
def first_second_third_highest(nums):
    first = None
    second = None
    third = None

    for num in nums:

        # Already present
        if num == first or num == second or num == third:
            continue

        # New highest
        if first is None or num > first:
            third = second
            second = first
            first = num

        # New second highest
        elif second is None or num > second:
            third = second
            second = num

        # New third highest
        elif third is None or num > third:
            third = num

    return first, second, third


def main():
    test_cases = [
        [10, 5, 20, 20, 8, 15, 3],
        [1, 2, 3],
        [10, 10, 10],
        [5, 4],
        [100],
        [],
        [-10, -5, -20, -1],
        [5, 5, 4, 4, 3, 3],
        [0, -1, -2, -3],
    ]

    for nums in test_cases:
        first, second, third = first_second_third_highest(nums)

        print(
            f"nums={nums}"
        )
        print(
            f"1st={first}, "
            f"2nd={second}, "
            f"3rd={third}"
        )
        print()


if __name__ == "__main__":
    main()
```

```python
import heapq


def top_k_largest(nums, k):
    if k <= 0 or not nums:
        return []

    if k >= len(nums):
        return sorted(nums, reverse=True)

    min_heap = []

    for num in nums:

        # Add current element
        heapq.heappush(min_heap, num)

        # Keep only K largest elements
        if len(min_heap) > k:
            heapq.heappop(min_heap)

    # Heap contains K largest,
    # but not necessarily sorted.
    return sorted(min_heap, reverse=True)


def main():
    test_cases = [
        ([10, 5, 20, 8, 15, 3, 25], 3),
        ([1, 2, 3, 4, 5], 2),
        ([5, 5, 5, 5], 2),
        ([], 3),
        ([10], 1),
        ([10], 5),
        ([1, 2, 3], 0),
        ([1, 2, 3], -1),
        ([-10, -5, -20, -1], 2),
    ]

    for nums, k in test_cases:
        result = top_k_largest(nums, k)

        print(
            f"nums={nums}, "
            f"k={k}, "
            f"top_k={result}"
        )


if __name__ == "__main__":
    main()
```

```python
import heapq


def top_k_smallest(nums, k):
    if k <= 0 or not nums:
        return []

    if k >= len(nums):
        return sorted(nums)

    # Python heapq is a min-heap.
    # Negate values to simulate a max-heap.
    max_heap = []

    for num in nums:
        heapq.heappush(max_heap, -num)

        if len(max_heap) > k:
            heapq.heappop(max_heap)

    result = [-x for x in max_heap]

    return sorted(result)


def main():
    test_cases = [
        ([10, 5, 20, 8, 15, 3, 25], 3),
        ([1, 2, 3, 4, 5], 2),
        ([-10, -5, -20, -1], 2),
        ([5, 5, 5, 5], 2),
        ([], 3),
        ([10], 1),
    ]

    for nums, k in test_cases:
        print(
            f"nums={nums}, "
            f"k={k}, "
            f"smallest={top_k_smallest(nums, k)}"
        )


if __name__ == "__main__":
    main()
```

```python
import heapq


def top_k_distinct(nums, k):
    if k <= 0:
        return []

    unique = set(nums)

    if len(unique) <= k:
        return sorted(unique, reverse=True)

    min_heap = []

    for num in unique:
        heapq.heappush(min_heap, num)

        if len(min_heap) > k:
            heapq.heappop(min_heap)

    return sorted(min_heap, reverse=True)


def main():
    test_cases = [
        ([10, 10, 9, 8, 8, 7], 3),
        ([5, 5, 5], 2),
        ([1, 2, 3, 4, 5], 3),
        ([], 3),
        ([10], 1),
        ([10, 20], 5),
    ]

    for nums, k in test_cases:
        print(
            f"nums={nums}, "
            f"k={k}, "
            f"top_k_distinct={top_k_distinct(nums, k)}"
        )


if __name__ == "__main__":
    main()
```

```python
# top k frquent element

from collections import Counter
import heapq


def top_k_frequent(nums, k):
    if k <= 0 or not nums:
        return []

    frequency = Counter(nums)

    min_heap = []

    for num, count in frequency.items():

        heapq.heappush(
            min_heap,
            (count, num)
        )

        if len(min_heap) > k:
            heapq.heappop(min_heap)

    return [
        num
        for count, num in sorted(
            min_heap,
            reverse=True
        )
    ]


def main():
    test_cases = [
        ([1, 1, 1, 2, 2, 3], 2),
        ([1], 1),
        ([1, 2, 3, 4], 2),
        ([5, 5, 5, 4, 4, 3], 2),
        ([], 2),
    ]

    for nums, k in test_cases:
        print(
            f"nums={nums}, "
            f"k={k}, "
            f"result={top_k_frequent(nums, k)}"
        )


if __name__ == "__main__":
    main()
```
