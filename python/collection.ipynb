# Frequency-based algorithms

```python
# Find the first non-repeating character.

from collections import Counter

def first_unique(s):
    count = Counter(s)

    for ch in s:
        if count[ch] == 1:
            return ch

    return None

print(first_unique("aabbcdde"))
```

# defaultdict(list) — grouping

Real-world applications
Group users by country

Group orders by customer

Group products by category

Group logs by service

Group employees by department

Build adjacency lists for graphs

```python
from collections import defaultdict

employees = [
    ("John", "IT"),
    ("Alex", "HR"),
    ("Bob", "IT"),
    ("Sam", "HR")
]

groups = defaultdict(list)

for name, department in employees:
    groups[department].append(name)

print(groups)
```

# OrderedDict
OrderedDict is a dictionary that provides specialized ordering operations.

Modern Python dictionaries already preserve insertion order, so OrderedDict is less important than it used

```python
from collections import OrderedDict

d = OrderedDict()

d["a"] = 1
d["b"] = 2
d["c"] = 3
d.move_to_end("a")
d
d.move_to_end("a", last=False)  # move to begining 
```

```python
# LRU

from collections import OrderedDict

class LRUCache:

    def __init__(self, capacity):
        self.capacity = capacity
        self.cache = OrderedDict()

    def get(self, key):
        if key not in self.cache:
            return -1

        self.cache.move_to_end(key)

        return self.cache[key]

    def put(self, key, value):

        if key in self.cache:
            self.cache.move_to_end(key)

        self.cache[key] = value

        if len(self.cache) > self.capacity:
            self.cache.popitem(last=False)
```

# ChainMap
combines multiple dictionaries into one logical view.

```python
from collections import ChainMap

defaults = {
    "color": "blue",
    "size": 10
}

user_config = {
    "color": "red"
}

config = ChainMap(user_config, defaults)

print(config["color"])
print(config["size"])
```
