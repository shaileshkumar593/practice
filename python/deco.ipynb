```python
def repeat(n):
    print(1)
    def decorator(func):
        print(2)
        def wrapper(*args, **kwargs):
            print(3)
            for _ in range(n):
                func(*args, **kwargs)
            print(4)
        return wrapper
    return decorator


@repeat(3)
def hello():
    print("Hello")


hello()
```
