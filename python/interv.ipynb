# Why is None used instead of [] as a default argument?

# Answer: Mutable default arguments (like lists, dictionaries, and sets) are created only once when the function is defined, so they are shared across all calls. Using None as the default and creating a new object inside the function ensures that each call gets its own fresh mutable object, preventing unexpected side effects.

```python
# correct 

def add(item, items=None):
    if items is None:
        items = []
    items.append(item)
    return items

print(add(1))
print(add(2))
print(add(3))
```

# Why does 100 is 100 evaluate to True?

# Answer: In CPython, small integers (typically from -5 to 256) are interned, meaning the interpreter reuses the same object for efficiency. As a result, two variables assigned the value 100 usually reference the same object, so is returns True. This is an implementation optimization and should not be relied on for numeric comparisons—use == instead.

# What is the difference between copy.copy() and copy.deepcopy()?

# Answer: copy.copy() creates a shallow copy, where only the outer object is copied and nested mutable objects are still shared. copy.deepcopy() recursively copies the entire object graph, creating independent copies of all nested objects, so changes to the copy do not affect the original.



```python
import copy

a = [[1],[2]]

b = copy.copy(a)

a[0].append(5)

print(b)

b.append(8)
print(b)
print(a)

a.append(80)

print(b)
print(a)
```

```python
import copy

a = [[1, 2], [3, 4]]

b = copy.deepcopy(a)

b[0][0] = 100

print(a)
print(b)
print(id(a))
print(id(b))

a.append(88)
b.append(55)
b[0][1]=67
a[0][0]=99

print(a)
print(b)
```

# What does return do inside a generator?

Answer: Inside a generator, return terminates the generator and raises a StopIteration exception. If a value is returned (for example, return 5), that value becomes the value attribute of the StopIteration exception (StopIteration.value). Normal iteration constructs like for loops catch and suppress this exception, so the returned value is not visible unless you catch StopIteration yourself.

```python
# closure 

def outer():

    x = 5

    def inner():
        print(x)

    return inner

f = outer()

f()
```

# Why does list(e) return only [(1, 'b')] after calling next(e)?

Answer: enumerate() returns an iterator, which produces items one at a time and remembers its current position. The call to next(e) consumes the first item (0, 'a'). When list(e) is called afterward, it continues from the current position and collects only the remaining item (1, 'b'). Once exhausted, the iterator cannot be reused
