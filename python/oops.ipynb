```python
class User:

    def __init__(self, name):
        self.name = name

    # getter 
    @property
    def name(self):
        return self._name

    @name.setter
    def name(self, value):
        if not value:
            raise ValueError("Name cannot be empty")

        self._name = value

    @name.deleter
    def name(self):
        del self._name
```

```python
u = User("Alice")

print(u.name)

u.name = "Bob"

del u.name
```

```python
class Rectangle:

    def __init__(self, width, height):
        self.width = width
        self.height = height

    @property
    def area(self):
        return self.width * self.height
```

```python
r = Rectangle(10, 5)

print(r.area)
```

```python
class Account:

    def __init__(self, balance):
        self.balance = balance

    @property
    def balance(self):
        return self._balance

    @balance.setter
    def balance(self, value):
        if value < 0:
            raise ValueError("Balance cannot be negative")

        self._balance = value
```

```python
account = Account(500)

account.balance = 1000
account.balance
```

```python
from abc import ABC, abstractmethod


class Employee(ABC):

    company = "TechCorp"

    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    @property
    def salary(self):
        return self._salary

    @salary.setter
    def salary(self, value):
        if value < 0:
            raise ValueError("Salary cannot be negative")

        self._salary = value

    @abstractmethod
    def calculate_bonus(self):
        pass

    def display(self):
        print(
            f"Name={self.name}, "
            f"Salary={self.salary}, "
            f"Company={self.company}"
        )


class Developer(Employee):

    def calculate_bonus(self):
        return self.salary * 0.10


class Manager(Employee):

    def calculate_bonus(self):
        return self.salary * 0.20


dev = Developer("Alice", 100000)
manager = Manager("Bob", 150000)

dev.display()
print(dev.calculate_bonus())

manager.display()
print(manager.calculate_bonus())
```

# Duck typing is Python's behavior-oriented approach where code generally cares whether an object supports the required operations rather than whether it belongs to a particular class or inheritance hierarchy. If an object provides the expected methods/behavior, it can be used.

```python
class Dog:
    def speak(self):
        return "Bark"


class Cat:
    def speak(self):
        return "Meow"


def speak(animal):
    return animal.speak()


print(speak(Dog()))  #  object implement behaviour 
print(speak(Cat()))
```

```python
from abc import ABC, abstractmethod


# ============================================================
# 1. ABSTRACTION
# ============================================================

class Payment(ABC):

    @abstractmethod
    def pay(self, amount):
        pass

    # Every concrete subclass must provide a property called payment_type.
    @property
    @abstractmethod
    def payment_type(self):
        pass


# ============================================================
# 2. INHERITANCE + ENCAPSULATION + PROPERTY
# ============================================================

class Employee:

    # Class variable
    company = "TechCorp"

    def __init__(self, name, salary):
        # Instance variables
        self.name = name
        self.salary = salary

    # PROPERTY - getter
    @property
    def salary(self):
        return self._salary

    # PROPERTY - setter
    @salary.setter
    def salary(self, value):

        if value < 0:
            raise ValueError("Salary cannot be negative")

        self._salary = value

    # Instance method
    def display(self):
        return f"{self.name} earns {self.salary}"

    # Class method
    @classmethod
    def get_company(cls):
        return cls.company

    # Static method
    @staticmethod
    def calculate_tax(salary):
        return salary * 0.10

    # Magic method
    def __str__(self):
        return f"Employee(name={self.name}, salary={self.salary})"


# ============================================================
# 3. INHERITANCE
# ============================================================

class Developer(Employee):

    def __init__(self, name, salary, language):
        # Calling parent constructor
        super().__init__(name, salary)

        self.language = language

    # METHOD OVERRIDING
    def display(self):
        return (
            f"Developer {self.name} "
            f"earns {self.salary} "
            f"and uses {self.language}"
        )

    # POLYMORPHIC METHOD
    def work(self):
        return f"{self.name} writes {self.language} code"


class Manager(Employee):

    def __init__(self, name, salary, team_size):
        super().__init__(name, salary)

        self.team_size = team_size

    def display(self):
        return (
            f"Manager {self.name} "
            f"earns {self.salary} "
            f"and manages {self.team_size} people"
        )

    def work(self):
        return f"{self.name} manages the team"


# ============================================================
# 4. PAYMENT IMPLEMENTATIONS
# ============================================================

class CreditCardPayment(Payment):

    @property
    def payment_type(self):
        return "Credit Card"

    def pay(self, amount):
        return f"Paid ₹{amount} using Credit Card"


class UPIPayment(Payment):

    @property
    def payment_type(self):
        return "UPI"

    def pay(self, amount):
        return f"Paid ₹{amount} using UPI"


# ============================================================
# 5. COMPOSITION
# ============================================================

class Engine:

    def start(self):
        return "Engine started"


class Car:

    def __init__(self):
        # Car HAS-A Engine
        self.engine = Engine()

    def start(self):
        return self.engine.start()


# ============================================================
# 6. DUCK TYPING
# ============================================================

class PostgreSQLRepository:

    def save(self, data):
        return f"Saved {data} to PostgreSQL"


class MongoRepository:

    def save(self, data):
        return f"Saved {data} to MongoDB"


def save_data(repository, data):
    # We don't care about the concrete class.
    # We only care that it has save().
    return repository.save(data)


# ============================================================
# 7. OPERATOR OVERLOADING
# ============================================================

class Money:

    def __init__(self, amount):
        self.amount = amount

    def __add__(self, other):
        return Money(self.amount + other.amount)

    def __str__(self):
        return f"₹{self.amount}"

    def __eq__(self, other):
        if not isinstance(other, Money):
            return NotImplemented

        return self.amount == other.amount


# ============================================================
# 8. __new__ + __init__
# ============================================================

class Product:

    def __new__(cls, name):
        print("__new__() → object is being created")

        return super().__new__(cls)

    def __init__(self, name):
        print("__init__() → object is being initialized")

        self.name = name

    def __str__(self):
        return f"Product: {self.name}"


# ============================================================
# MAIN
# ============================================================

def main():

    print("\n========== CLASS & OBJECT ==========")

    developer = Developer(
        "Alice",
        100000,
        "Python"
    )

    manager = Manager(
        "Bob",
        150000,
        5
    )

    print(developer)
    print(manager)


    print("\n========== PROPERTY ==========")

    print("Salary:", developer.salary)

    developer.salary = 120000

    print("Updated Salary:", developer.salary)

    # This will raise ValueError
    # developer.salary = -500


    print("\n========== CLASS VARIABLE ==========")

    print(Employee.company)
    print(developer.company)
    print(manager.company)


    print("\n========== CLASS METHOD ==========")

    print(Employee.get_company())


    print("\n========== STATIC METHOD ==========")

    tax = Employee.calculate_tax(100000)

    print("Tax:", tax)


    print("\n========== INHERITANCE ==========")

    print(developer.display())
    print(manager.display())


    print("\n========== POLYMORPHISM ==========")

    employees = [
        developer,
        manager
    ]

    for employee in employees:
        print(employee.work())


    print("\n========== ABSTRACTION ==========")

    payments = [
        CreditCardPayment(),
        UPIPayment()
    ]

    for payment in payments:
        print(payment.payment_type)
        print(payment.pay(1000))


    print("\n========== DUCK TYPING ==========")

    postgres = PostgreSQLRepository()
    mongo = MongoRepository()

    print(save_data(postgres, {"id": 1}))
    print(save_data(mongo, {"id": 2}))


    print("\n========== COMPOSITION ==========")

    car = Car()

    print(car.start())


    print("\n========== OPERATOR OVERLOADING ==========")

    money1 = Money(100)
    money2 = Money(200)

    money3 = money1 + money2

    print("Money 1:", money1)
    print("Money 2:", money2)
    print("Money 1 + Money 2:", money3)

    print(
        "Equality:",
        money3 == Money(300)
    )


    print("\n========== __new__ vs __init__ ==========")

    product = Product("Laptop")

    print(product)


if __name__ == "__main__":
    main()
```
