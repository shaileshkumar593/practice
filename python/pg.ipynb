# Class Variable vs Instance Variable

```python
class User:
    roles = []


u1 = User()
u2 = User()

u1.roles.append("admin")

print(u2.roles)
```

Why?

roles is a class attribute.

             User
              │
          roles = []
           /     \
          ↓       ↓
         u1      u2
Both access the same list.

```python
class User:

    def __init__(self):
        self.roles = []
```

u1.roles → separate list
u2.roles → separate list

But class variables are not always bad
Immutable/shared configuration is often appropriate:

class User:
    MAX_LOGIN_ATTEMPTS = 5
That's intentionally shared.

The trap is unintentionally sharing mutable state.

# MRO + super()

MRO determines the order Python uses to resolve methods in inheritance hierarchies. Python uses C3 linearization. super() doesn't simply mean “call the parent”; it delegates to the next implementation according to the MRO, which is particularly important in multiple inheritance.

```python
class A:
    def hello(self):
        print("A")


class B(A):
    def hello(self):
        print("B")


class C(A):
    def hello(self):
        print("C")


class D(B, C):
    pass
```

```python
D().hello()
```

```python
print(D.mro())
```

# Object-Sharing Argument Semantics

```python
def change(x):
    x = 100


a = 10

change(a)

print(a)
```

Why?

Inside the function:

x → 10
Then:

x = 100
rebinds the local variable x.

It doesn't rebind a.

```python
def change(items):
    items.append(100)


a = [1, 2, 3]

change(a)

print(a)
```

# Async Generator vs Coroutine

```python
async def users():
    yield 1
    yield 2
    yield 3


async for user in users():
    print(user)
```

```python
async def get_user():
    return {"id": 1}

result = get_user()  #  create coroutine object 
print(result)

user = await get_user()  # consume coroutine 
print(user)
```

```python
async def telemetry():
    while True:
        data = await receive_data()
        yield data


"""
    This is excellent for:

    MQTT telemetry;

    WebSocket streams;

    streaming HTTP responses;

    async DB records;

    LLM token streaming.


"""
async for data in telemetry():
    print(data)
```
